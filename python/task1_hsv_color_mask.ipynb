```python
import cv2
import numpy as np
import urllib.request
from matplotlib import pyplot as plt
url = "https://platform.theverge.com/wp-content/uploads/sites/2/chorus/uploads/chorus_asset/file/7172985/tconnors_160823_1227_0316.0.jpg"
req = urllib.request.urlopen(url)
arr = np.asarray(bytearray(req.read()), dtype=np.uint8)
img_bgr = cv2.imdecode(arr, -1)
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.title('Original Robot Image')
plt.axis('off')
plt.show()
```

```python
img_hsv = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2HSV)

# --- GÖREVİNİZ BURADA BAŞLIYOR ---
# Mavi renk için doğru HSV numaralarını bulun ve sıfırları/255'leri değiştirin..
lower_limit = np.array([100, 100, 150])
upper_limit = np.array([130, 255, 255])
# --- GÖREVİNİZ BURADA BİTİYOR ---

mask = cv2.inRange(img_hsv, lower_limit, upper_limit)
plt.imshow(mask, cmap='gray')
plt.title('Filtered Mask')
plt.axis('off')
plt.show()
```
